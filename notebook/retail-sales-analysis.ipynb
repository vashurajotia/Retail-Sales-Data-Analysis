# Exploratory Data Analysis (EDA)

## 1. Key Performance Indicators (KPIs)

```python
total_sales = df_clean['Sales'].sum()
total_profit = df_clean['Profit'].sum()
total_orders = df_clean['Order ID'].nunique()
total_quantity = df_clean['Quantity'].sum()

average_order_value = total_sales / total_orders
profit_margin = (total_profit / total_sales) * 100
print("total sales: ", round(total_sales, 2))
print("total profit: ", round(total_profit, 2))
print("total orders: ",total_orders)
print("total quantity sold: ",total_quantity)
print("Average order value: ", round(average_order_value, 2))
print("profit margin: ", round(profit_margin, 2),"%")
```

```python
category_sales = df_clean.groupby('Category')['Sales'].sum().sort_values(ascending = False)
print(category_sales)
```

```python
plt.figure(figsize = (9,5))
category_sales.plot(kind = 'bar')

plt.title("Sales by Product Category")
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation = 0)

plt.show()
```

```python
region_sales = df_clean.groupby('Region')['Sales'].sum().sort_values(ascending = False)
print(region_sales)
```

```python
plt.figure(figsize = (9,5))
sns.barplot(x = region_sales.index, y = region_sales.values)

plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')

plt.show()
```

## Top 10 Best-Selling Products

```python
top_products = (
    df_clean.groupby('Product Name')['Sales']
    .sum()
    .sort_values(ascending = False)
    .head(10)
)
print(top_products)
```

```python
plt.figure(figsize = (12,6))

top_products.sort_values().plot(kind = 'barh')

plt.title('Top 10 products by sales')
plt.xlabel('Total Sales')
plt.ylabel('Product Name')

plt.show()
```

## Monthly Sales Trend Analysis

```python
df_clean["Month"] = df_clean['Order Date'].dt.to_period('M')
monthly_sales = df_clean.groupby('Month')['Sales'].sum()
print(monthly_sales)
```

```python
plt.figure(figsize=(14,6))

monthly_sales.index = monthly_sales.index.astype(str)

plt.plot(monthly_sales.index, monthly_sales.values, marker = 'o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.xticks(rotation =90)
plt.grid(True)

plt.show()
```

## Profit Analysis

```python
category_profit = (df_clean.groupby("Category")['Profit'].sum().sort_values(ascending=False))
print(category_profit)
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(x=category_profit.index, y=category_profit.values)

plt.title('Profit by product category')
plt.xlabel('category')
plt.ylabel('Total Profit')

plt.show()
```

## Advanced Bussiness Analysis

### 1.Discount Analysis

```python
discount_analysis = df_clean.groupby('Discount').agg(
    Total_sales = ('Sales','sum'),
    Total_Profit = ('Profit','sum'),
    Total_Orders = ('Order ID', 'nunique')
)
print(discount_analysis)
```

### 2. Subcategory Profitability

```python
subcategory_profit =(
    df_clean.groupby('Sub-Category')['Profit'].sum().sort_values())
print(subcategory_profit)
```

```python
plt.figure(figsize = (12,7))

subcategory_profit.plot(kind = 'barh')

plt.title('Profit by product subcategory')
plt.xlabel('Total Profit')
plt.ylabel('Subcategory')

plt.tight_layout()
plt.show()
```

### 3. Loss-Making Products

```python
loss_product = (df_clean.groupby('Product Name')['Profit'].sum().sort_values().head(10))
print(loss_product)
```

### 4. Regional Profitability

```python
regional_profit = df_clean.groupby('Region').agg(
    Sales = ('Sales', 'sum'),
    Profit = ('Profit', 'sum')
)

regional_profit['Profit_Margin'] = (
    regional_profit['Profit']/
    regional_profit['Sales'] * 100
)

print(regional_profit.sort_values(
    by = 'Profit', ascending = False))
```
